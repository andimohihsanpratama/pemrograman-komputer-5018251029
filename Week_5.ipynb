```python
# List berisi minimal 20 nilai sensor (contoh data, termasuk beberapa nilai negatif)
sensor_data = [
    45, 62, 33, -5, 78, 50, 41, 88, 65, 30,
    -2, 55, 72, 48, 90, 39, 60, 44, 82, 51, 25, 79
]

# Tentukan batas ambang (threshold) untuk status ALARM
THRESHOLD_ALARM = 75

print("=== PROGRAM SENSOR LOG ANALYZER ===\n")
print(f"{'Sampel':<10} | {'Nilai Sensor':<15} | {'Status':<10}")
print("-" * 43)

valid_count = 0
total_nilai = 0
jumlah_alarm = 0
nilai_min = None
nilai_max = None

# 1. Menampilkan nomor sampel dan nilai, serta 2. Menentukan status NORMAL/ALARM
for i, nilai in enumerate(sensor_data, start=1):
    # 6. Mengabaikan nilai negatif sebagai data tidak valid
    if nilai < 0:
        print(f"{i:<10} | {nilai:<15} | TIDAK VALID (Negatif)")
        continue

    valid_count += 1
    total_nilai += nilai

    # Menentukan status NORMAL atau ALARM
    if nilai > THRESHOLD_ALARM:
        status = "ALARM"
        # 4. Menghitung jumlah alarm
        jumlah_alarm += 1
    else:
        status = "NORMAL"

    print(f"{i:<10} | {nilai:<15} | {status}")

    # 5. Mencari nilai minimum dan maksimum secara eksplisit (tanpa min()/max())
    if nilai_min is None or nilai < nilai_min:
        nilai_min = nilai
    if nilai_max is None or nilai > nilai_max:
        nilai_max = nilai

print("\n" + "=" * 43)
print("=== RINGKASAN ANALISIS SENSOR ===")

if valid_count > 0:
    # 3. Menghitung rata-rata secara eksplisit (tanpa sum())
    rata_rata = total_nilai / valid_count

    print(f"Total Data Valid      : {valid_count} sampel")
    print(f"Rata-rata Nilai       : {rata_rata:.2f}")
    print(f"Total Kejadian Alarm  : {jumlah_alarm}")
    print(f"Nilai Minimum         : {nilai_min}")
    print(f"Nilai Maksimum        : {nilai_max}")
else:
    print("Tidak ada data valid yang dapat dianalisis.")
```
