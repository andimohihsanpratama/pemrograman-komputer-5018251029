#Soal 1
Analisis Output Program elif (Pemahaman Urutan Kondisi) Perhatikan potongan kode Python berikut: score = 85 if score >= 90: grade = "A" elif score >= 80: grade = "B" elif score >= 70: grade = "C" else: grade = "D" print(f"Grade Anda: {grade}")

Pertanyaan:

- Berapa output yang akan dicetak oleh program di atas?

- Jelaskan mengapa Python mengevaluasi nilai tersebut sebagai grade tersebut berdasarkan sifat pengecekan dari atas ke bawah!

```python
Score = 85.0
if Score >= 90:
    Grade = "A"
elif Score >= 80:
    Grade = "B"
elif Score >= 70:
    Grade = "C"
else:
    Grade = "D"

print(Grade)
```

#Soal 2
Validasi Rentang Data (Range Check) Sensor Sebuah sistem monitoring industri memerlukan validasi data tekanan (pressure). Tekanan sensor dianggap Valid jika berada di dalam rentang 10.0 hingga 50.0 (inklusif/termasuk angka 10.0 dan 50.0). Jika di luar itu, sistem mencetak "Out of Range".

Pertanyaan:

- Tulislah potongan kode Python menggunakan chained comparison (perbandingan berantai) yang efisien untuk menyelesaikan kasus ini!

- Uji kode Anda menggunakan variabel pressure = 45.5

```python
pressure = 45.5

if 10.0 <= pressure <= 50.0:
    print("Valid")
else:
    print("Out of Range")
```

#Soal 3
Operator Logika (and / or) untuk Sistem Peringatan Sebuah mesin kapal akan memicu status "ALARM" jika memenuhi dua kondisi secara bersamaan:

1. Temperatur mesin melebihi 80.0 derajat.

2. Status pompa air pendingin (pump_running) bernilai False (mati).

Pertanyaan:

- Buatlah blok kode percabangan menggunakan operator logika yang tepat untuk mendeteksi kondisi tersebut!

- Jika diketahui temperature = 85.0 dan pump_running = False, apa output yang dihasilkan?

```python
temperature = 85.0
pump_running = False

if temperature > 80.0 and not pump_running:
    print("ALARM")
else:
    print("NORMAL")
```

#Soal 4
Studi Kasus Klasifikasi Level Bahaya (Threshold-Based Monitoring) Buatlah sebuah program Python yang meminta pengguna memasukkan nilai pengukuran zat kimia (x), lalu mengklasikannya ke dalam ketentuan berikut:

- 0 <= x < 40: Status "LOW"

- 40 <= x < 70: Status "NORMAL"

- 70 <= x < 90: Status "HIGH"

- 90 <= x <= 100: Status "CRITICAL"

Selain rentang tersebut: Status "INVALID"

Catatan: Sertakan f-string untuk memperjelas tampilan output akhir.

```python
x = float(input("Masukkan nilai pengukuran zat kimia (x): "))

if 0 <= x < 40:
    status = "LOW"
elif 40 <= x < 70:
    status = "NORMAL"
elif 70 <= x < 90:
    status = "HIGH"
elif 90 <= x <= 100:
    status = "CRITICAL"
else:
    status = "INVALID"

print(f"Hasil Pengukuran: {x} -> Status: {status}")
```
